## 3. Avaliação

Não existe um gabarito rotulado de “avaliações relevantes” para cada pergunta, então usamos **métricas proxy**
baseadas nos metadados, que **não** são usados na recuperação (a busca é puramente semântica sobre o texto):

* **Coerência de sentimento**: para perguntas sobre problemas, a % das evidências com nota 1–2 deveria ser alta;
  para perguntas sobre elogios, a % com nota 4–5.
* **Coerência temática (entrega)**: para perguntas sobre entrega, a % de evidências de pedidos atrasados ou não
  entregues deveria ser muito maior que a taxa da base.
* **Fundamentação**: % das evidências citadas na resposta e quantidade de citações inválidas.

Também comparamos o pipeline **com e sem a transformação da pergunta** (multi-query).

```python
docs = pd.read_parquet(config.DOCS_PATH)
base_neg = (docs.review_score <= 2).mean(); base_pos = (docs.review_score >= 4).mean()
base_entrega = (docs.delivered_late.eq(True) | (docs.order_status != "delivered")).mean()
print(f"Taxas na base: nota 1-2 = {base_neg:.0%} | nota 4-5 = {base_pos:.0%} | atrasado ou não entregue = {base_entrega:.0%}")

AVAL = {
    "Quais são os principais problemas relatados pelos clientes em suas avaliações?": "neg",
    "Quais padrões podem ser identificados nas avaliações de clientes insatisfeitos?": "neg",
    "Existem padrões recorrentes nas avaliações relacionadas à qualidade dos produtos?": None,
    "Quais aspectos da experiência de compra são mais elogiados pelos clientes?": "pos",
    "O que os clientes mais elogiam sobre a entrega?": "pos",
    "O que os clientes relatam sobre problemas relacionados à entrega?": "entrega",
    "Quais avaliações sustentam a conclusão de que existem problemas recorrentes relacionados ao prazo de entrega?": "entrega",
    "Quais reclamações relacionadas à entrega aparecem com maior frequência?": "entrega",
}

def metricas(evs, alvo):
    m = pd.DataFrame([e.metadata for e in evs])
    if m.empty: return {}
    atraso = m.get("delivered_late", pd.Series(dtype=object)).eq(True) | (m["order_status"] != "delivered")
    return {"evidencias": len(m), "sim_media": round(sum(e.similarity for e in evs) / len(evs), 3),
            "%nota_1_2": round((m.review_score <= 2).mean() * 100), "%nota_4_5": round((m.review_score >= 4).mean() * 100),
            "%atraso_ou_nao_entregue": round(atraso.mean() * 100)}

linhas = []
rag_sem = VoCRAG(use_query_rewriting=False)
for q, alvo in AVAL.items():
    for nome, sistema in [("com multi-query", rag), ("sem multi-query", rag_sem)]:
        r = respostas[q] if (sistema is rag and q in respostas) else sistema.ask(q)  # reaproveita respostas da seção 1
        linhas.append({"pergunta": q[:70], "alvo": alvo, "pipeline": nome, "status": r.status,
                       **metricas(r.evidences, alvo), "%citadas": round(len(r.cited_refs) / max(len(r.evidences), 1) * 100),
                       "citacoes_invalidas": len(r.invalid_refs)})
aval = pd.DataFrame(linhas)
aval
```

```python
def acerto(row):
    return {"neg": row["%nota_1_2"], "pos": row["%nota_4_5"], "entrega": row["%atraso_ou_nao_entregue"]}.get(row["alvo"])
aval["coerencia_%"] = aval.apply(acerto, axis=1)
resumo = aval.dropna(subset=["coerencia_%"]).pivot_table(index="alvo", columns="pipeline", values="coerencia_%", aggfunc="mean").round(1)
resumo["taxa_na_base_%"] = pd.Series({"neg": base_neg, "pos": base_pos, "entrega": base_entrega}).mul(100).round(1)
resumo
```

```python
print("Médias gerais por pipeline:")
aval.groupby("pipeline")[["evidencias", "sim_media", "%citadas", "citacoes_invalidas"]].mean().round(2)
```

```python
aval.groupby(["pipeline", "status"]).size().unstack(fill_value=0)
```

### Leitura dos resultados

* A coerência das evidências com o tema da pergunta fica **muito acima da taxa da base**, o que indica que a
  recuperação semântica encontra avaliações realmente relacionadas (lembrando que os metadados não foram usados na busca).
* As respostas citam uma parcela grande das evidências e não há citações inválidas — os insights são
  **rastreáveis** até o `review_id` da avaliação.
* Os testes da seção 2 mostram as três guardas em ação: o sistema prefere declarar a falta de evidências a gerar uma
  resposta sem suporte.

**Limitações conhecidas**
* “Frequência” é estimada a partir da vizinhança semântica (até 1.000 avaliações), não de uma contagem exaustiva.
* Avaliações muito curtas (“ok”, “bom”) carregam pouco contexto; elas são deduplicadas mas permanecem na base.
* O viés de quem escreve comentários (clientes insatisfeitos comentam mais) deve ser considerado na interpretação.
